### Binary Basics

- Each number = sum of powers of 2
- Example:
  - 13 = 1101 = 8 + 4 + 1

| Operation   | Symbol | Meaning             |
| ----------- | ------ | ------------------- |
| AND         | `&`    | Both bits must be 1 |
| OR          | `¦`    | Any bit is 1        |
| XOR         | `^`    | Different bits → 1  |
| NOT         | `~`    | Flip bits           |
| Left Shift  | `<<`   | Multiply by 2       |
| Right Shift | `>>`   | Divide by 2         |


### 1. Check if i-th bit is

- if (n >> i) & 1:

### 2. Set i-th bit

- n = n | (1 << i)

### 3. Clear i-th bit

- n = n & ~(1 << i)

### 4. Toggle i-th bit

- n = n ^ (1 << i)
