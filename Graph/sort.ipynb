## Topological Sorting (for DAG only)

DAG = Directed Acyclic Graph

- Used when tasks have dependencies.
- Course scheduling (LeetCode 207)


## 1. BFS-Based Topological Sort (Kahn’s Algorithm)

🧠 Intuition

- Find nodes with 0 dependencies (indegree = 0)
- These can be done first
- Remove them from graph
- Update others


```python
from collections import deque

class Graph:
    def __init__(self):
        self.graph = {}
        self.indegree = {}

    def add_edge(self, u, v):#u:independent, v: dependent on u
        if u not in self.graph:
            self.graph[u] = []
        if v not in self.graph:
            self.graph[v] = []

        if u not in self.indegree:
            self.indegree[u] = 0
        if v not in self.indegree:
            self.indegree[v] = 0

        self.graph[u].append(v)
        self.indegree[v] += 1

    def topo_sort_bfs(self):
        queue = deque([node for node in self.graph if self.indegree[node] == 0])
        result = []

        while queue:
            node = queue.popleft()
            result.append(node)

            for neighbor in self.graph[node]:
                self.indegree[neighbor] -= 1
                if self.indegree[neighbor] == 0:
                    queue.append(neighbor)

        return result
```

## 2. DFS-Based Topological Sort (Stack Method)

- Instead of: “Who can I do first?” (BFS)
- We think: “Finish dependencies first, then push current node”
- Key Idea: Postorder DFS = Topological Order (reversed)


```python
class GraphDFS:
    def __init__(self):
        self.graph = {}

    def add_edge(self, u, v):
        if u not in self.graph:
            self.graph[u] = []
        if v not in self.graph:
            self.graph[v] = []

        self.graph[u].append(v)

    def topo_sort_dfs(self):
        visited = set()
        stack = []

        def dfs(node):
            visited.add(node)

            for neighbor in self.graph[node]:
                if neighbor not in visited:
                    dfs(neighbor)

            stack.append(node)  # POSTORDER

        for node in self.graph:
            if node not in visited:
                dfs(node)

        return stack[::-1]  # reverse
    
    def topo_sort_dfs_with_cycle(self):
        visited = set()
        path = set()  # recursion stack
        stack = []

        def dfs(node):
            if node in path:
                return False  # cycle detected

            if node in visited:
                return True

            path.add(node)
            visited.add(node)

            for neighbor in self.graph[node]:
                if not dfs(neighbor):
                    return False

            path.remove(node)
            stack.append(node)
            return True

        for node in self.graph:
            if node not in visited:
                if not dfs(node):
                    return []  # cycle exists

        return stack[::-1]
            

```
